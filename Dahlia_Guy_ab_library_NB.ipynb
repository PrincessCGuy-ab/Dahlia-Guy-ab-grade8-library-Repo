```python
import json
with open("/content/books.json", "r") as file:
 data = json.load(file)
print("JSON file loaded")
ref = db.reference("books")
for books in data["books"]:
 ref.child(str(books["id"])).set(books)
print("Data uploaded successfully!")
```

```python
books_ref = db.reference("books")
books_data = books_ref.get()
print(books_data)
```

```python
for book_details in books_data.values():
  print(book_details["title"])
```

```python
for book_details in books_data.values():
  print(f"Title: {book_details['title']}, Available: {book_details['details']['available']}")
```

```python
total_copies = 0

for book in books_data.values():
  total_copies += book['details']['copies']

print(f"Total number of books: {total_copies}")
```
